### 2) Eliminación de Columnas Irrelevantes

```python
df = df.drop(columns=["ClienteID"])

# Verificamos
df.info()
```

#### 3.2) Clasificación de tus variables categóricas

```python
# Convertir "Genero" a numérico

df["Genero"] = df["Genero"].map({"Female": 0, "Male": 1}) # Convertir a numérico

df["Genero"].value_counts() 
```

```python
# Convertir "Lineas_Multiples" a numérico

df["Lineas_Multiples"] = df["Lineas_Multiples"].replace("No phone service", "No")
df["Lineas_Multiples"] = df["Lineas_Multiples"].map({"No": 0, "Yes": 1})

df["Lineas_Multiples"].value_counts()
```

```python
# Convertir servicios de internet a numérico

cols_internet = [
    "Seguridad_Online",
    "Backup_Online",
    "Proteccion_Dispositivo",
    "Soporte_Tecnico",
    "Streaming_TV",
    "Streaming_Peliculas"
]

for col in cols_internet:
    df[col] = df[col].replace("No internet service", "No")
    df[col] = df[col].map({"No": 0, "Yes": 1})
```

```python
df = pd.get_dummies(
    df,
    columns=["Servicio_Internet", "Tipo_Contrato", "Metodo_Pago"],
    drop_first=True
) # Convertir variables categóricas a dummies
```

```python
df.info() # Verificar que todo esté en formato numérico
```

```python
df.head() # Verificar la transformación final
```

```python
# Convertir columnas booleanas a numéricas

bool_cols = df.select_dtypes(include="bool").columns
df[bool_cols] = df[bool_cols].astype(int)
```
