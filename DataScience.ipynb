## 🛠️ Preparacion de los Datos

### 1) Cargar el dataset

```python
import pandas as pd

# Cargar el archivo
df = pd.read_csv("telecomx_clean.csv")

# Verificar estructura
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df["Churn"].value_counts()
```

### 2) Eliminación de Columnas Irrelevantes

```python
df = df.drop(columns=["ClienteID"])

# Verificamos
df.info()
```

### 3) Embeddings

#### 3.1) Analisis de variables 

```python
df.select_dtypes(include="object").columns
```

```python
for col in df.select_dtypes(include="object").columns:
    print(col)
    print(df[col].unique())
    print("-----")
```

#### 3.2) Clasificación de tus variables categóricas

```python
# Convertir "Genero" a numérico

df["Genero"] = df["Genero"].map({"Female": 0, "Male": 1}) # Convertir a numérico

df["Genero"].value_counts() 
```

```python
# Convertir "Lineas_Multiples" a numérico

df["Lineas_Multiples"] = df["Lineas_Multiples"].replace("No phone service", "No")
df["Lineas_Multiples"] = df["Lineas_Multiples"].map({"No": 0, "Yes": 1})

df["Lineas_Multiples"].value_counts()
```

```python
# Convertir servicios de internet a numérico

cols_internet = [
    "Seguridad_Online",
    "Backup_Online",
    "Proteccion_Dispositivo",
    "Soporte_Tecnico",
    "Streaming_TV",
    "Streaming_Peliculas"
]

for col in cols_internet:
    df[col] = df[col].replace("No internet service", "No")
    df[col] = df[col].map({"No": 0, "Yes": 1})
```

```python
df = pd.get_dummies(
    df,
    columns=["Servicio_Internet", "Tipo_Contrato", "Metodo_Pago"],
    drop_first=True
) # Convertir variables categóricas a dummies
```

```python
df.info() # Verificar que todo esté en formato numérico
```

```python
df.head() # Verificar la transformación final
```

```python
# Convertir columnas booleanas a numéricas

bool_cols = df.select_dtypes(include="bool").columns
df[bool_cols] = df[bool_cols].astype(int)
```

```python
df.info()
```

## 🎯 Correlación y Selección de Variables

### 1) Análisis de Correlación

```python
df.info()
```

```python
# Visualización de la matriz de correlación

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(14,10))
sns.heatmap(df.corr(), cmap="coolwarm", center=0)
plt.title("Matriz de Correlación")
plt.show()
```

```python
# Analizar correlación con la variable objetivo "Churn"

corr_churn = df.corr()["Churn"].sort_values(ascending=False)
print(corr_churn)
```

```python
df[["Antiguedad_Meses","Cargo_Mensual","Cargo_Total"]].corr() 
# Analizar correlación entre variables numéricas
```

### 2) Análisis Dirigido

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
sns.boxplot(x="Churn", y="Antiguedad_Meses", data=df)
plt.title("Antigüedad vs Cancelación")
plt.show() # Visualizar relación entre antigüedad y cancelación
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="Churn", y="Cargo_Total", data=df)
plt.title("Cargo Total vs Cancelación")
plt.show() # Visualizar relación entre cargo total y cancelación
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(
    x="Antiguedad_Meses",
    y="Cargo_Total",
    hue="Churn",
    alpha=0.6,
    data=df
)
plt.title("Antigüedad vs Cargo Total por Churn")
plt.show()
```
